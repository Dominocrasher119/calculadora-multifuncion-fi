## Entrada y validación de datos
Código procedente de `utils.py`.


```python
# Grupo: C12 - Calculadora Multifunción
# Integrantes: Iker Sánchez Prieto, Roger Pedra Vilalta y Bernat Marín López

import math


class EntradaSalida:
    """Lee y valida entradas sin guardar estado ni requerir una instancia.

    Los métodos estáticos se llaman directamente desde otros módulos.
    """

    @staticmethod
    def tecladoInt(mensaje="Ingrese un número entero: "):
        """Lee un entero; repite la pregunta hasta recibir una entrada válida."""
        while True:
            # int convierte el texto; si no es un entero, se vuelve a preguntar.
            try:
                return int(input(mensaje))
            except ValueError:
                # Capturar el error evita cerrar el programa por una entrada incorrecta.
                print("Entrada inválida. Escribe un número entero.")

    @staticmethod
    def tecladoFloat(mensaje="Ingrese un número (puede usar coma o punto decimal): "):
        """Lee un número finito y acepta coma o punto como separador decimal."""
        while True:
            texto = input(mensaje).strip()
            try:
                # float() entiende el punto; se convierte la coma antes de validar.
                numero = float(texto.replace(",", "."))
                if math.isfinite(numero):
                    return numero
                print("Entrada inválida. El número debe ser finito.")
            except ValueError:
                print("Número incompleto o inválido. Ejemplos: 12, 3.5 o 3,5.")

    @staticmethod
    def tecladoString():
        """Lee una cadena que no esté vacía ni contenga solo espacios."""
        while True:
            texto = input("Ingrese un texto: ").strip()
            if texto:
                return texto
            print("El texto no puede estar vacío.")

    @staticmethod
    def tecladoChar():
        """Lee exactamente un carácter, ignorando espacios exteriores."""
        while True:
            texto = input("Ingrese un carácter: ").strip()
            if len(texto) == 1:
                return texto
            print("Debe ingresar un solo carácter.")
```

## Operaciones aritméticas
Código procedente de `operaciones.py`.


```python
# Grupo: C12 - Calculadora Multifunción
# Integrantes: Iker Sánchez Prieto, Roger Pedra Vilalta y Bernat Marín López

# Archivo Python para implementar operaciones aritméticas básicas.


class CalculadoraAritmetica:
    """Opera con dos valores numéricos, a y b, y devuelve el resultado.

    En la resta y la división importa el orden de los valores.
    """

    def suma(self, a, b):
        """Devuelve la suma de a y b."""
        return a + b

    def resta(self, a, b):
        """Devuelve la resta de b a a."""
        return a - b

    def multiplicacion(self, a, b):
        """Devuelve el producto de a y b."""
        return a * b

    def division(self, a, b):
        """Devuelve a dividido entre b; lanza ValueError si b es cero."""
        # La división entre cero no está definida; el menú captura el error.
        if b == 0:
            raise ValueError("No se puede dividir entre cero.")
        return a / b
```

## Media ponderada
Código procedente de `notas.py`.


```python
# Grupo: C12 - Calculadora Multifunción
# Integrantes: Iker Sánchez Prieto, Roger Pedra Vilalta y Bernat Marín López

es = EntradaSalida

"""Cálculo interactivo de la media ponderada de varias notas."""

import math



def calcular_media_ponderada(notas, pesos):
    """Devuelve la media ponderada tras validar notas y pesos.

    Cada nota se empareja con el peso de la misma posición, que indica su
    importancia en la media. Las notas deben estar entre 0 y 10. Los pesos deben ser
    no negativos y su suma debe ser mayor que cero.

    Raises:
        ValueError: Si las listas están vacías, no tienen igual longitud o
            contienen valores no válidos.
    """
    if not notas or not pesos:
        raise ValueError("Debe proporcionarse al menos una nota y su peso.")
    if len(notas) != len(pesos):
        raise ValueError("Debe haber el mismo número de notas y pesos.")

    suma_ponderada = 0
    suma_pesos = 0

    # Se recorre cada pareja y se acumulan nota por peso y suma de pesos.
    for nota, peso in zip(notas, pesos):
        try:
            nota = float(nota)
            peso = float(peso)
        except (TypeError, ValueError):
            raise ValueError("Las notas y los pesos deben ser números válidos.")

        if not math.isfinite(nota) or nota < 0 or nota > 10:
            raise ValueError("Cada nota debe ser un número entre 0 y 10.")
        if not math.isfinite(peso) or peso < 0:
            raise ValueError("Cada peso debe ser un número no negativo.")

        suma_ponderada += nota * peso
        suma_pesos += peso

    # El total debe ser positivo para poder dividir y obtener la media.
    if suma_pesos == 0:
        raise ValueError("La suma de los pesos debe ser mayor que cero.")

    return suma_ponderada / suma_pesos


def calcular_media_ponderada_interactiva():
    """Lee varias notas y sus pesos, muestra el resultado y permite repetir."""
    while True:
        cantidad = es.tecladoInt("¿Cuántas notas quieres introducir?: ")
        while cantidad <= 0:
            print("Debes introducir al menos una nota.")
            cantidad = es.tecladoInt("¿Cuántas notas quieres introducir?: ")

        notas = []
        pesos = []

        # Se repite la lectura de cada pareja para asegurar que cumple los rangos.
        for numero in range(1, cantidad + 1):
            nota = es.tecladoFloat(f"Nota {numero} (0-10): ")
            while nota < 0 or nota > 10:
                print("La nota debe estar entre 0 y 10.")
                nota = es.tecladoFloat(f"Nota {numero} (0-10): ")

            peso = es.tecladoFloat(f"Peso de la nota {numero}: ")
            while peso < 0:
                print("El peso no puede ser negativo.")
                peso = es.tecladoFloat(f"Peso de la nota {numero}: ")

            notas.append(nota)
            pesos.append(peso)

        try:
            resultado = calcular_media_ponderada(notas, pesos)
            print(f"\nLa nota media ponderada es: {resultado:.2f}\n")
        except ValueError as error:
            # Por ejemplo, si todos los pesos son cero, se informa y se continúa.
            print(f"\nNo se pudo calcular la media: {error}\n")

        repetir = input("¿Quieres calcular otra media ponderada? (s/n): ").strip().lower()
        while repetir not in ("s", "si", "sí", "n", "no"):
            repetir = input("Responde s o n: ").strip().lower()
        if repetir in ("n", "no"):
            break
```

## Potencias y logaritmos
Código procedente de `calculos.py`.


```python
# Grupo: C12 - Calculadora Multifunción
# Integrantes: Iker Sánchez Prieto, Roger Pedra Vilalta y Bernat Marín López

"""Funciones de cálculo de ingeniería"""

import math


def potencia_dos(exponente):
    """Devuelve 2 elevado a exponente, incluso si es decimal o negativo.

    Valida que sea finito y controla el desbordamiento; ** calcula la potencia.
    """
    try:
        exponente = float(exponente)
    except (TypeError, ValueError):
        raise ValueError("El exponente debe ser un número válido.")
    if not math.isfinite(exponente):
        raise ValueError("El exponente debe ser un número finito.")
    try:
        return 2 ** exponente
    except OverflowError:
        raise ValueError("El resultado es demasiado grande para representarlo.")


def logaritmo_base_dos(numero):
    """Devuelve el exponente al que elevar 2 para obtener numero.

    Valida que el número sea positivo y finito y utiliza math.log2.
    """
    try:
        numero = float(numero)
    except (TypeError, ValueError):
        raise ValueError("El número debe ser válido.")
    if not math.isfinite(numero) or numero <= 0:
        raise ValueError("El número debe ser positivo y finito.")
    # math.log2 no depende del intervalo arbitrario [-100, 100] de la búsqueda
    # binaria original y mantiene precisión para valores muy grandes o pequeños.
    return math.log2(numero)
```

## Conversiones de bases y unidades
Código procedente de `conversiones.py`.


```python
# Grupo: C12 - Calculadora Multifunción
# Integrantes: Iker Sánchez Prieto, Roger Pedra Vilalta y Bernat Marín López

"""Funciones para convertir bases numéricas y unidades informáticas."""

import math


_BASES = {
    "binario": 2,
    "octal": 8,
    "decimal": 10,
    "hexadecimal": 16,
}

# Los factores expresan cada unidad en KiB; cada salto multiplica por 1024.
_UNIDADES_KIB = {
    "KiB": 1,
    "MiB": 1024,
    "GiB": 1024 * 1024,
}


def convertir_base(numero, base_origen, base_destino):
    """Convierte un número entero entre binario, octal, decimal y hexadecimal.

    ``numero`` se recibe como texto para conservar dígitos como los ceros
    iniciales y las letras hexadecimales. Se admiten valores negativos.
    Devuelve un entero para destino decimal y texto para las demás bases.
    Una base desconocida o un dígito inválido provoca ValueError.
    """
    origen = str(base_origen).strip().lower()
    destino = str(base_destino).strip().lower()
    if origen not in _BASES or destino not in _BASES:
        raise ValueError("La base debe ser binario, octal, decimal o hexadecimal.")

    texto = str(numero).strip()
    # int interpreta y valida los dígitos según la base de origen.
    try:
        valor = int(texto, _BASES[origen])
    except (TypeError, ValueError):
        raise ValueError("El número no es válido para la base de origen.")

    if destino == "decimal":
        return valor
    # format representa el entero en la base elegida y conserva su signo.
    return format(valor, {"binario": "b", "octal": "o", "hexadecimal": "X"}[destino])


def convertir_unidades(cantidad, origen, destino):
    """Convierte cantidades entre KiB, MiB y GiB usando múltiplos de 1024.

    Se admiten cantidades decimales y negativas. Los valores no finitos y las
    unidades no reconocidas producen un ``ValueError`` controlado.
    """
    if origen not in _UNIDADES_KIB or destino not in _UNIDADES_KIB:
        raise ValueError("La unidad debe ser KiB, MiB o GiB.")
    try:
        cantidad = float(cantidad)
    except (TypeError, ValueError):
        raise ValueError("La cantidad debe ser numérica.")
    if not math.isfinite(cantidad):
        raise ValueError("La cantidad debe ser un número finito.")

    # Primero se expresa la cantidad en KiB y después en la unidad de destino.
    resultado = cantidad * _UNIDADES_KIB[origen] / _UNIDADES_KIB[destino]
    if resultado.is_integer():
        return int(resultado)
    return resultado
```

## Menús y navegación
Código procedente de `menu.py`.


```python
# Grupo: C12 - Calculadora Multifunción
# Integrantes: Iker Sánchez Prieto, Roger Pedra Vilalta y Bernat Marín López

es = EntradaSalida


# El menú conecta las entradas con los módulos de cálculo y conversión.
# Las opciones del menú se corresponden con su posición en estas listas.
menuGeneral = [
    "Operaciones aritméticas",
    "Cálculo de media ponderada",
    "Conversión de bases numéricas",
    "Conversión de unidades informáticas",
    "Cálculo de potencias de 2",
    "Logaritmos en base 2",
]
submenuAritmetica = ["Suma", "Resta", "Multiplicación", "División"]


def mostrar_titulo(titulo):
    """Imprime un título centrado para distinguir las secciones de la consola."""
    print("\n" + "=" * 42)
    print(titulo.center(42))
    print("=" * 42)


def printMenu(menu, textoOpcionCero="Salir", titulo="Menú"):
    """Muestra las opciones y devuelve una selección válida.

    Args:
        menu: Lista de opciones que se van a mostrar.
        textoOpcionCero: Texto asociado a la opción 0.
        titulo: Título que aparece encima de las opciones.

    Returns:
        La opción elegida, entre 0 y la cantidad de elementos del menú.
    """
    mostrar_titulo(titulo)
    for numero, opcion in enumerate(menu, start=1):
        print(f"  {numero}. {opcion}")
    print(f"  0. {textoOpcionCero}\n")

    # El bucle mantiene el menú abierto hasta que se introduce una opción del rango.
    opcion = es.tecladoInt("Selecciona una opción: ")
    while opcion < 0 or opcion > len(menu):
        print("Opción inválida. Inténtalo de nuevo.\n")
        opcion = es.tecladoInt("Selecciona una opción: ")
    return opcion


def ejecutar_operacion(calculadoraAritmetica, opcion):
    """Pide dos números, ejecuta una operación y permite repetirla.

    Args:
        calculadoraAritmetica: Instancia que realiza el cálculo.
        opcion: Número de la operación seleccionada en el submenú.
    """
    nombre = submenuAritmetica[opcion - 1].lower()
    mostrar_titulo(nombre.capitalize())

    while True:
        numero1 = es.tecladoFloat("  Primer número: ")
        numero2 = es.tecladoFloat("  Segundo número: ")

        # Cada rama selecciona el método asociado a la opción.
        try:
            if opcion == 1:
                resultado = calculadoraAritmetica.suma(numero1, numero2)
            elif opcion == 2:
                resultado = calculadoraAritmetica.resta(numero1, numero2)
            elif opcion == 3:
                resultado = calculadoraAritmetica.multiplicacion(numero1, numero2)
            elif opcion == 4:
                resultado = calculadoraAritmetica.division(numero1, numero2)
            else:
                print("Opción de operación inválida.\n")
                return
            print(f"\n  Resultado: {resultado}\n")
        except ValueError as error:
            # La división avisa si el divisor es cero; se informa sin salir.
            print(f"\n  Error: {error}\n")

        # Se valida la respuesta antes de repetir o volver al submenú.
        repetir = input("¿Quieres hacer otra operación del mismo tipo? (s/n): ").strip().lower()
        while repetir not in ("s", "si", "sí", "n", "no"):
            repetir = input("Responde s o n: ").strip().lower()
        if repetir in ("n", "no"):
            print()
            break
        print(f"\n¡Seguimos con la {nombre}!\n")


def repetir_calculo():
    """Pregunta si se quiere repetir una operación de ingeniería."""
    repetir = input("¿Quieres hacer otro cálculo? (s/n): ").strip().lower()
    while repetir not in ("s", "si", "sí", "n", "no"):
        repetir = input("Responde s o n: ").strip().lower()
    return repetir in ("s", "si", "sí")


def ejecutar_calculo_ingenieria(opcion):
    """Gestiona la entrada, el cálculo y la repetición de las opciones 5 y 6."""
    while True:
        try:
            if opcion == 5:
                exponente = es.tecladoFloat("Exponente de 2: ")
                resultado = potencia_dos(exponente)
                print(f"\n2 elevado a {exponente:g} = {resultado:g}\n")
            else:
                numero = es.tecladoFloat("Número positivo: ")
                resultado = logaritmo_base_dos(numero)
                print(f"\nLogaritmo en base 2 de {numero:g} = {resultado:g}\n")
        except ValueError as error:
            print(f"\nError: {error}\n")
        if not repetir_calculo():
            break


def ejecutar_conversion_bases():
    """Solicita los datos y permite repetir una conversión de base."""
    bases = {"2": "binario", "8": "octal", "10": "decimal", "16": "hexadecimal"}
    while True:
        numero = input("Número entero que deseas convertir: ").strip()
        origen = input("Base de origen (binario, octal, decimal, hexadecimal): ").strip().lower()
        destino = input("Base de destino (binario, octal, decimal, hexadecimal): ").strip().lower()
        origen = bases.get(origen, origen)
        destino = bases.get(destino, destino)
        try:
            resultado = convertir_base(numero, origen, destino)
            print(f"\nResultado: {resultado}\n")
        except ValueError as error:
            print(f"\nError: {error}\n")
        if not repetir_calculo():
            break


def ejecutar_conversion_unidades():
    """Solicita datos y permite repetir una conversión de unidades."""
    unidades = {"kib": "KiB", "mib": "MiB", "gib": "GiB"}
    while True:
        cantidad = es.tecladoFloat("Cantidad: ")
        origen = input("Unidad de origen (KiB, MiB, GiB): ").strip().lower()
        destino = input("Unidad de destino (KiB, MiB, GiB): ").strip().lower()
        origen = unidades.get(origen, origen)
        destino = unidades.get(destino, destino)
        try:
            resultado = convertir_unidades(cantidad, origen, destino)
            print(f"\nResultado: {resultado} {destino}\n")
        except ValueError as error:
            print(f"\nError: {error}\n")
        if not repetir_calculo():
            break


def iniciar_calculadora():
    """Inicia la aplicación y mantiene la navegación hasta que el usuario salga."""
    calculadora = CalculadoraAritmetica()
    mostrar_titulo("CALCULADORA MULTIFUNCIÓN")

    # El menú principal se repite hasta que se elige salir.
    while True:
        opcion = printMenu(menuGeneral, titulo="Menú principal")
        if opcion == 0:
            print("\n¡Hasta pronto!\n")
            break
        elif opcion == 1:
            # El submenú se repite hasta elegir 0, que devuelve al menú principal.
            while True:
                operacion = printMenu(
                    submenuAritmetica,
                    textoOpcionCero="Volver",
                    titulo="Operaciones aritméticas",
                )
                if operacion == 0:
                    break
                ejecutar_operacion(calculadora, operacion)
        elif opcion == 2:
            calcular_media_ponderada_interactiva()
        elif opcion == 3:
            ejecutar_conversion_bases()
        elif opcion == 4:
            ejecutar_conversion_unidades()
        elif opcion == 5 or opcion == 6:
            ejecutar_calculo_ingenieria(opcion)
        else:
            # Rama de seguridad: printMenu limita la selección al rango válido.
            print("\nFunción no válida\n")
```
